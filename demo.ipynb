```python
# Упр. 1

df = pd.read_csv('https://raw.githubusercontent.com/MVRonkin/BasicDataAnalysisCourse/refs/heads/main/Workshops/cars.csv')

df_selet = df.copy()
df_selet = df_selet[
    (df['Price(euro)'] < 90000) &\
    (df['Engine_capacity(cm3)'].between(200, 5000, inclusive='neither')) &\
    (df['Distance'].between(5000, 500000)) &\
    (~(df['Year'] < 1990))
].drop_duplicates().reset_index()

df_selet.info()

df_drop = df_selet.copy()
df_drop = df_drop[
    (df_drop['Price(euro)'] >= 90000) &\
    (df_drop['Engine_capacity(cm3)'])
]
```

```python
df_nulls = df_selet.copy()
df_nulls = df_nulls.dropna(subset=['Make', 'Model'])
removed = df_selet.loc[df_selet.index.difference(df_nulls.index)]
df_nulls
```
